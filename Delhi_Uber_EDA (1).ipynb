FOR FINDING THE OUTLIERS

```python
booking_values = pd.to_numeric(df['Booking Value'], errors='coerce')
upper_booking_value = booking_values.mean() + 3 * booking_values.std()
lower_booking_value = booking_values.mean() - 3 * booking_values.std()
booking_value_outliers = df[(booking_values > upper_booking_value) | (booking_values < lower_booking_value)]
booking_value_outliers
df['Booking Value']
```
